```python
import sys
import json
import os

try:
    import stanza
except ImportError:
    print("Please install stanza: pip install stanza torch")
    sys.exit(1)
```

```python
stanza.download('vi')
p = stanza.Pipeline('vi', processors='tokenize,pos,lemma,depparse')

input_file = 'output/clauses.txt'
output_file = 'output/dependency.json'

if not os.path.exists(input_file):
    print(f"Error: Could not find {input_file}")
    sys.exit(1)

with open(input_file, 'r', encoding='utf-8') as f:
    lines = f.readlines()
lines
```

```python
results = []
clause_id = 0

for line in lines:
    text = line.strip()
    if not text:
        continue
        
    clause_id += 1
    print(f"Processing clause {clause_id}: {text}")
    
    doc = p(text)
    
    dependencies = []
    token_offset = 0

    for sent in doc.sentences:
        for word in sent.words:
            tok_idx = word.id + token_offset
            
            tok_text = word.text
            
            head_local = word.head
            if head_local == 0:
                head_idx = 0
                head_text = "ROOT"
            else:
                head_idx = head_local + token_offset
                head_w = next((hw for hw in sent.words if hw.id == head_local), None)
                if head_w:
                    head_text = head_w.text
                else:
                    head_text = "UNKNOWN"
            
            dep_rel = word.deprel
            
            dependencies.append({
                "token_idx": tok_idx,
                "token": tok_text,
                "head_idx": head_idx,
                "head": head_text,
                "dep": dep_rel
            })
        if sent.words:
            token_offset += sent.words[-1].id
        
    results.append({
        "clause_id": clause_id,
        "text": text,
        "dependencies": dependencies
    })
```

```python
os.makedirs(os.path.dirname(output_file), exist_ok=True)
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"Successfully processed {clause_id} clauses.")
print(f"Dependency output saved to {output_file}")
```
